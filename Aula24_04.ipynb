```python
#Escrever um programa que recebe um numero e retorna os números pares.
num= int(input("Digite um número: "))
for i in range (num+1):
  if i % 2 == 0:
    print (i)
```

```python
#Escrever um programa que recebe um número e imprime ao contrario
num = int(input("Digite um número: "))
for i in range (num+ 1):
  print (num - i)
```

```python
palavra = str(input("Digite uma palavra:"))
palindromo = ""
for i in (palavra):
  palindromo = i+palindromo
print(palindromo)
```

```python
#Escrever um programa que recebe um numero de 1 a 10 e imprimi a tabua
num = int(input("Digite fale um número :"))
if num >= 0 and num <= 10:
  for i in range (0, 11):
    print (f"{i} * {num}  = {i * num}")
else:
  print ("Número inválido")
```

```python
#Par ou ímpar
#Peça um número ao usuário e informe se ele é par ou ímpar.
num= int(input("Diga seu número"))
par = num % 2
if par == 0:
  print ("Esse número é par")
else:
  print ("Esse número é impar")
```

```python
#Maior de dois números
#Solicite dois números e mostre qual é o maior (ou se são iguais).
num1= float(input("diga um número"))
num2= float(input("diga um número"))

if num1 == num2:
  print ("Esses números são iguais")
elif num1 > num2:
  print ("primeiro número é maior doque o segundo")
elif num1 < num2:
  print ("Segundo número é maior doque o primeiro")
```

```python
#Peça a nota de um aluno e informe:
#o Aprovado (nota ≥ 7)
#o Recuperação (5 ≤ nota &lt; 7)
#o Reprovado (nota &lt; 5)

nota = float(input("Qual a sua nota?"))
if nota >= 5 and nota < 7:
  print ("Você está de recuperação")
elif nota >= 7:
  print ("Você está aprovado")
elif nota < 5:
  print ("Você está reprovado.")
```

```python
#Receba a idade e classifique:
#o Criança (0–12)
#o Adolescente (13–17)
#o Adulto (18–59)
#o Idoso (60+)

id= int(input("Qual a sua idade?"))
if id <= 12 and id > 0:
  print ("Você é criança")
elif id >=13 and id <= 17:
  print ("Voce é adolescente")
elif id >= 18 and id <= 59:
  print ("Você é adulto")
elif id >= 60:
  print ("Você é idoso")
```

```python
#Defina uma senha fixa no código. Peça ao usuário para digitar e informe se está
#correta ou incorreta.
senha = 1234
ts = int(input("Digite a senha")) #tentativa da senha

if ts == senha:
  print ("Senha correta")
else:
  print("Senha incorreta")
```

```python
#Peça um número e diga se ele é positivo, negativo ou zero.
num = float(input("Digite um número"))
if num == 0:
  print ("Esse número é o zero")
elif num > 0:
  print ("esse número é postivio")
elif num < 0:
  print ("Esse número é negativo")
```

```python
#Peça dois números e uma operação (+, -, *, /). Use if/elif para calcular o
#resultado.
print("Opção= 1 Soma                  ")
print("Opção= 2 Subtração             ")
print("Opção= 3 Divisão               ")
print("Opção= 4 Multiplicação         ")
print("")

num1 = float(input("Digite o primeiro numero"))
op= int(input("Escolha a opção desejada"))
num2 = float(input("Digite o segundo número"))

if op == 1:
  print(f"A soma é igual a {num1 + num2}")
elif op ==2:
  print(f"A subtração é igual a {num1 - num2}")
elif op ==3:
  print (f"A divisão é igual a {num1 / num2}")
elif op==4:
  print (f"A multiplicação é igual a {num1 * num2}")
else:
  print ("Opção inválida")
```

```python
#Peça um número de 1 a 7 e mostre o dia correspondente (ex: 1 = domingo).
dia = int(input("Escolha um número e direi qual é o dia da semana."))
if dia == 1:
  print("Domingo")
elif dia== 2:
  print("Segunda-feira")
elif dia==3:
  print("Terça-feira")
elif dia==4:
  print("Quarta-feira")
elif dia==5:
  print("Quinta-feira")
elif dia==6:
  print("Sexta-feira")
elif dia==7:
  print("Sábado")
else:
  print("número indisponível")
```

```python
#Escolha um ano e decida se é bissexto ou não
ano = int(input("Digite um ano"))
bi = ano % 4
bis = ano % 100
bise = ano % 400

if bi == 0 or bise == 0:
  print ("Esse ano é bissexto")
else:
  print("Esse ano não é bissexto")
```

```python
#Peça o valor de uma compra:
# ≥ 100: 10% de desconto
# ≥ 50 e &lt; 100: 5% de desconto
# &lt; 50: sem desconto
val= float(input("Digite o valor da compra"))
if val >= 100:
  print("A sua compra tem 10% de desconto")
elif val <= 100 and val >= 50:
  print("A sua compra terá 5% de desconto")
elif val < 50:
  print("A sua compra não terá desconto")
```

```python
matriz_2x2 = [
    [1, 2],
    [3, 4]
]

print("Matriz 2x2:")
for linha in matriz_2x2:
    print(*linha)
```

```python
matriz_2x3=[
    [1, 2, 3],
    [4, 5, 6]
]
print ("Matriz 2x3:")
for linha in matriz_2x3:
  print (*linha)
```

```python
matriz_nomes=[
    ["Ana", "João"],
    ["Maria", "Pedro"]
]
print ("Matriz de nomes: ")
for linha in matriz_nomes:
  print(*linha)
```

```python
matriz_3x3=[
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
print ("Matriz 3x3:")
for linha in matriz_3x3:
  print (*linha)
```

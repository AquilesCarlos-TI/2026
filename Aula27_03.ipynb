```python
#Estrutura de decisões (if, elif e else)
#Escrever um programa que compara 2 números

#entrada

a= int(input("Digite um numero inteiro"))
b= int(input("Digite um numero inteiro"))

#processamento/ saida

if a == b:
  print(f"{a} é igual a {b}")
else:
  print(f"{a} não é igual a {b}")
```

```python
#Descubra se um número é par ou impar
numero= int(input("Digite um número"))

if numero % 2 == 1:
  print(f"esse número ({numero}) é impar")
else:
  print(f"esse número ({numero}) é par")
```

```python
#idade > 12= adolescente
#idade < 12= criança
#idade > 18= adulto

nome= str(input("Digite o seu nome"))
id= int(input("Digite a sua idade"))

if id <= 12 and id >=1:
  print(f"seu nome é {nome} e você é criança")
elif id >=13 and id <= 17:
  print (f"seu nome é {nome} e você é adolescente")
elif id >=18:
  print(f"seu nome é {nome} e voce é adulto")
else:
  print("idade inválida")
```

```python
#escrever um progaram ade uma calculadora simples. O usuário ira digitar 2 numeros e irá escolher uma operação aritimética.
#entrada
print("----------CALCULADORA----------")
print("                               ")
print("-------------MENU--------------")
print("                               ")
print("Opção= 1 Soma                  ")
print("Opção= 2 Subtração             ")
print("Opção= 3 Divisão               ")
print("Opção= 4 Multiplicação         ")
print("                               ")
print("-------------------------------")

num1 = float(input("Digite o primeiro numero"))
op= int(input("Escolha a opção desejada"))
num2 = float(input("Digite o segundo número"))

if op == 1:
  print(f"A soma é igual a {num1 + num2}")
elif op ==2:
  print(f"A subtração é igual a {num1 - num2}")
elif op ==3:
  print (f"A divisão é igual a {num1 / num2}")
elif op==4:
  print (f"A multiplicação é igual a {num1 * num2}")
else:
  print ("Opção inválida")
```

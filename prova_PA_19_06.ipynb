```python
#Solicite ao usuário um número inteiro positivo e calcule seu fatorial utilizando um laço de repetição.

#Exemplo:

#Digite um número: 5

#Fatorial = 120

#Pois:

#5! = 5 × 4 × 3 × 2 × 1

#Observação:
#O programa deve validar se o número informado é positivo.
num= int(input("Diga um número"))
fatorial=1
for i in range (1, num+1):
  fatorial= (fatorial*i)
print (fatorial)
```

Solicite ao usuário o valor do saque e informe a menor
quantidade possível de notas para realizar o pagamento.

Exemplo:

Digite o valor do saque: 287

Saída esperada:

2 nota(s) de R$100
1 nota(s) de R$50
1 nota(s) de R$20
1 nota(s) de R$10
1 nota(s) de R$5
1 nota(s) de R$2

Requisitos:
- Utilizar laços de repetição.
- Não utilizar listas.
- Exibir apenas as notas utilizadas.

```python
saque= int(input("Qual valor deseja sacar?"))
cem= 0
cinq= 0
vin= 0
dez= 0
cin= 0
dois=0
um= 0
while saque != 0:
  if saque >=100:
    saque -= 100
    cem +=1
  elif saque >= 50:
    saque -= 50
    cinq +=1
  elif saque >= 20:
    saque -= 20
    vin +=1
  elif saque >= 10:
    saque -= 10
    dez +=1
  elif saque >= 5:
    saque -= 5
    cin +=1
  elif saque >= 2:
    saque -= 2
    dois +=1
  elif saque >=1:
    saque -= 1
    um +=1
print (f"{cem} notas de R$100 ")
print (f"{cinq} notas de R$50 ")
print (f"{vin} notas de R$20")
print (f"{dez} notas de R$10")
print (f"{cin} notas de R$5 ")
print (f"{dois} notas de R$2")
print (f"{um} notas de R$1")
```

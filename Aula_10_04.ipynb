```python
#input
numero = int(input("Digite um número e descubra se é par."))
num = (numero % 2)
#output and process
if num == 0:
  print ("Esse número é par.")
elif num == 1:
  print("Esse número é ímpar.")
else:
  print("Número inválido.")
```

```python
#input
nota1 = float(input("Digite a nota do 1° bimestre"))
nota2 = float(input("Digite a nota do 2° bimestre"))
nota3 = float(input("Digite a nota do 3° bimestre"))
nota4 = float(input("Digite a nota do 4° bimestre"))
media = ((nota1 + nota2 + nota3 + nota4) / 4)

#processamento
if media >= 7 and media < 10:
  print ("você passou")
elif media <=7 and media >= 5:
  print ("Você tem que fazer recuperação")
elif media < 5 and media > 0:
  print ("Você não passou de ano.")
elif media < 0 or media > 10:
  print ("Notas inválidas")
```

```python
# até 100 sem desconto
#até 500 mais de 101 10%
#mais de 500 20%

#input
valor= float(input("Digite o valor de sua compra"))

#process
if valor < 100 and valor > 0:
  print (f"Compra sem desconto, o valor será: {valor}")
elif valor >= 100 and valor < 500:
  print (f"Compra com desconto de 10%, o valor final será {(valor) - (valor * 0.1)}")
elif valor >= 500:
  print (f"Compra com desconto de 20%, o valor final será {(valor) - (valor * 0.2)}")
else:
  print (f"O valor: {valor} é inválido")
```
